## 3. Sequential Coordination

In sequential coordination, agents work one after another. Each agent builds on the output of the previous one.

**Use case:** Research → Analysis → Final Recommendation

```python
def sequential_research(company: str):
    print(f"Sequential Research for {company}")

    step1 = f"Step 1 - Researcher: Gathered overview for {company}."   #calling agent 1
    step2 = f"Step 2 - Analyst: Analyzed profit drivers and AI projects for {company}." #calling agent 2
    step3 = f"Step 3 - Reviewer: Final recommendation for {company} based on previous steps." #calling agent 3

    return f"{step1}\n{step2}\n{step3}"

print(sequential_research("Microsoft"))
```

## 4. Parallel Task Execution

In parallel execution, multiple agents work simultaneously on different aspects and then combine results.

**Use case:** Multiple agents research different aspects at the same time.

```python
import concurrent.futures

def parallel_research(company: str): #parallel_agents
    print(f"Parallel Research for {company}")

    def researcher1(): return f"AI Projects for {company}" # inside this functino define your first agent
    def researcher2(): return f"SDG Alignment for {company}" # inside this functino define your second agent
    def researcher3(): return f"Profit Drivers for {company}" # inside this functino define your third agent

    with concurrent.futures.ThreadPoolExecutor() as executor:
        results = list(executor.map(lambda f: f(), [researcher1, researcher2, researcher3]))

    return " | ".join(results)

print(parallel_research("NVIDIA"))
```

## 5.  Review and Critique Loops

In review and critique loops, one agent generates output and another critiques it. This improves quality through iteration.

```python
def review_and_critique(company: str):
    print(f"Review & Critique Loop for {company}")

    draft = f"Draft analysis for {company}."
    critique = f"Critique: Draft is good but needs more SDG details for {company}."
    revised = f"Revised analysis for {company} after critique."

    return f"Draft: {draft}\nCritique: {critique}\nRevised: {revised}"

print(review_and_critique("Google"))
```

## 6. Comparison of Communication Strategies

**Sequential Coordination:**
- Simple and easy to understand
- Good for linear workflows
- Slower because steps happen one after another

**Parallel Task Execution:**
- Faster execution
- Good when tasks are independent
- Requires good result merging logic

**Review and Critique Loops:**
- Highest quality output
- Good for complex analysis
- More expensive in tokens and time

**Production Recommendation:**
- Use **Sequential** for simple tasks
- Use **Parallel** for speed
- Use **Review & Critique** for high-stakes decisions
- Most real systems use a **hybrid** approach

## 7. Final Test Query – Compare All Three Strategies
