2. Core Streamlit Components
Streamlit provides various input components that allow users to interact with your app. Let's explore the most common ones:

 2.1 Basic Input Components
First, let's import streamlit:


## Key Streamlit Commands
- st.title() → Displays a title
- st.text_input() → Creates a text input field
- st.write() → Displays text or other content

````python
st.title("Text Input Example")

name = st.text_input("What's your name?", "Type here...")

if name != "Type here...":
    st.write(f"Hello {name}!")

# Password input
password = st.text_input("Enter password", type="password")
````

### Number Input

````python
st.title("Number Input Example")

age = st.number_input(
    "Enter your age",
    min_value=0,
    max_value=120,
    value=25
)

st.write(f"You are {age} years old")
````

### Button

````python
st.title("Button Example")

if st.button("Click me!"):
    st.write("Button was clicked!")
else:
    st.write("Button hasn't been clicked yet")
````

## 2.2 Selection Components

### Slider

````python
st.title("Slider Examples")

# Numeric slider
number = st.slider("Pick a number", 0, 100, 50)

st.write(f"You picked: {number}")
````

````python
from datetime import time

appointment = st.slider(
    "Schedule your appointment:",
    value=(time(11, 30), time(12, 45))
)

st.write("Your appointment:", appointment)
````

### Checkbox

````python
st.title("Checkbox Example")

if st.checkbox("Show/Hide"):
    st.write("You can see this text!")
````

### Radio Buttons

````python
st.title("Selection Examples")

favorite_color = st.radio(
    "What's your favorite color?",
    ["Red", "Green", "Blue"]
)

st.write(f"Your favorite color is {favorite_color}")
````

# 3. Displaying Data in Streamlit

````python
st.title("Text Display Examples")

# Using st.write()
st.write("Normal text with st.write()")

st.write("You can write", "multiple", "arguments")

# Using st.markdown()
st.markdown("# This is a heading")

st.markdown("**Bold text** and *italic text*")

st.markdown("- Bullet point 1\n- Bullet point 2")
````

````python
st.title("Data Structure Examples")

# List example
my_list = ["apple", "banana", "orange"]

st.write("List:", my_list)

# Dictionary example
my_dict = {
    "name": "John",
    "age": 30,
    "city": "New York"
}

st.write("Dictionary:", my_dict)
````

````python
import pandas as pd

df = pd.DataFrame({
    'Name': ['John', 'Anna', 'Peter'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Paris', 'London']
})

st.write("Simple table:")
st.table(df)

st.write("Interactive dataframe:")
st.dataframe(df)
````

## 3.3 File Uploader

````python
st.title("File Uploader Example")

uploaded_file = st.file_uploader(
    "Choose a CSV file",
    type="csv"
)

if uploaded_file is not None:
    df = pd.read_csv(uploaded_file)

    st.write("Data from your CSV file:")
    st.dataframe(df)
````

# 4. Creating Interactive Applications

````python
st.title("Counter Example with Session State")

if 'count' not in st.session_state:
    st.session_state.count = 0

if st.button('Increment'):
    st.session_state.count += 1

st.write('Count = ', st.session_state.count)
````

## 4.2 Forms

````python
st.title("Contact Form Example")

with st.form("contact_form"):

    name = st.text_input("Name")

    email = st.text_input("Email")

    message = st.text_area("Message")

    submitted = st.form_submit_button("Submit")

    if submitted:
        st.success("Thank you for your message!")

        st.write(f"We'll contact {name} at {email}")
````
