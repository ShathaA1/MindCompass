##### Approach 2 : Read Scanned PDF

````python
import pdf2image
from PIL import Image
import pytesseract
import os

pdf_path = '/content/resume-sample.pdf'
detected_text = ''

# Make sure the Tesseract executable is in your system's PATH.
# If you're on Windows, you might need to specify the path manually.
# pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'

# Check if the PDF file exists
if not os.path.exists(pdf_path):
    print(f"Error: The file '{pdf_path}' was not found.")
else:
    # Use pdf2image to convert each page of the PDF into an image
    images = pdf2image.convert_from_path(pdf_path)

    # Loop through each image and use pytesseract to get the text
    for i, page_image in enumerate(images):
        print(f"Extracting text from page {i + 1}...")

        # Use pytesseract to perform OCR (Optical Character Recognition)
        # This is where the magic happens! It "reads" the text from the image.
        page_text = pytesseract.image_to_string(page_image)
        detected_text += page_text + '\n\n'

    print("Done extracting text from all pages.")

# Print the final combined text
print("\n--- Extracted Text ---")
print(detected_text)
````

#### Step 3 : Continuing the Conversation

Often, we would want to have conversations with the LLM which are more than just a pair of a single prompt and a single response. Let us now learn how to use our past conversation history to continue the conversation.

To simplify the implementation, we define the following function for calling the OpenAI GPT API from now on -

````python
def continue_chat(system_message, user_assistant_messages,client):

    # The system message is  formatted as a list with one dictionary
    system_msg = [{"role": "system", "content": system_message}]

    # This creates a list of messages with alternating 'user' and 'assistant' roles
    # based on their position in the list. This logic is still correct.
    user_assistant_msgs = [
        {"role": "assistant", "content": user_assistant_messages[i]}
        if i % 2
        else {"role": "user", "content": user_assistant_messages[i]}
        for i in range(len(user_assistant_messages))
    ]

    # Combine the system message and the chat history
    allmsgs = system_msg + user_assistant_msgs

    # The new API uses `client.chat.completions.create()`
    # We set the model to "gpt-4.1-mini"
    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=allmsgs
    )

    # The response is now an object, so we use dot notation to access its attributes
    return response.choices[0].message.content
````

The function accepts -

* system_message (string) : This acts as the system_msg
* user_assistant_messages (list) : This list contains user prompts and model responses in alternating order. This is also the order in which they occur in the conversation.

The function internally makes the API call to generate and return a new response based on the conversation history.

Let us now use this function to continue our previous conversation, and find out the highest paying jobs out of the ones recommended in the first response.

We will use the same system message (system_msg) used in previous call.

We create user_assistant_messages list as follows -

````python
user_msg1 = user_msg
model_response1 = response.choices[0].message.content
user_msg2 = 'based on the suggestions, choose the 3 jobs with highest average salary'
user_assistant_msgs = [user_msg1, model_response1, user_msg2]
````

Note that we used the original prompt as the first user message (user_msg1), the response to that prompt as the first model response message (model_response1), and our new prompt as the second user message (user_msg2).

Finally, we add them to the user_assistant_messages list in order of their occurrence in the conversation.

We now call the continue_chat() function to get the next response in the conversation.

````python
response = continue_chat(system_msg, user_assistant_msgs,client)
````

So far, we've used the RetrievalQA chain, a LangChain type for pulling document pieces from a vector store and asking one question about them. But, sometimes we need to have a full conversation about a document, including referring to topics we've already talked about.

Thankfully, LangChain has us covered. To make this possible, our system needs a memory or conversation history.  We will show the manual way of doing it with RetrievalQA first

````python
# Assuming the retriever and chat history are already set up


# 1. Set up the LLM and the retriever
llm = ChatOpenAI(
    model="gpt-4.1-mini",
    temperature=0,
    openai_api_key=os.environ.get("OPENAI_API_KEY")
)

# Assuming vector_index is loaded and embeddings are passed
# vector_index = FAISS.load_local('index_store', embeddings)
# retriever = vector_index.as_retriever(search_kwargs={"k": 6})

# Initialize an empty chat history list
chat_history = []

# 2. First question: 'in what context is Morocco mentioned in the report?'
# This is a basic retrieval query
query1 = "in what context is Morocco mentioned in the report?"

# Use the simple RetrievalQA chain you're already familiar with
qa_interface = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=retriever
)

# Get the answer for the first query
result1 = qa_interface.invoke(query1)
print("Answer to first query:", result1)

# Add the first question and answer to the chat history
chat_history.append((query1, result1))

print("-" * 50)

# 3. Second question: 'give some statistics around this.'
# This query is dependent on the previous answer.
# To make it conversational, we combine the chat history with the new query.
query2 = "give some statistics around this."

# Create a new, combined query that includes the context of the conversation
# This is the key step to handle conversation simply
combined_query = f"Chat History: {chat_history}\nUser Query: {query2}"

# Get the answer for the second, conversational query
result2 = qa_interface.invoke(combined_query)
print("Answer to second query:", result2)
````

Let's ask the PDF to reveal the context in which Morocco is mentioned in the report.

'chat_history' parameter is a list contains past conversation history. For the first message, this list will be empty.

'question' parameter is used to send our message.

### Better way of doing this with LCEL.
LCEL, or the LangChain Expression Language, is the new standard for building complex LLM applications. It's a key part of LangChain's evolution, moving from monolithic, pre-built chains to a modular system.

The benefit of this LCEL approach is that each component is a separate and testable part of the pipeline. This modularity is a core principle of LangChain's new design.

More info here [link text](https://python.langchain.com/docs/concepts/lcel/)

````python
# No need to re-import libraries here; this is for demonstration.
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_classic.chains.retrieval import create_retrieval_chain
from langchain_classic.chains.history_aware_retriever import create_history_aware_retriever

# 1. Setup the LLM and the retriever
llm = ChatOpenAI(
    model="gpt-4.1-mini",
    openai_api_key=os.environ.get("OPENAI_API_KEY")
)

embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small",
    openai_api_key=os.environ.get("OPENAI_API_KEY")
)

vector_index = FAISS.load_local(
    'index_store',
    embeddings,
    allow_dangerous_deserialization=True
)
retriever = vector_index.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 6}
)

# 2. Define the LCEL prompt and document chain
prompt = ChatPromptTemplate.from_template("""
Answer the following question based only on the provided context:

<context>
{context}
</context>

Question: {input}""")

document_chain = create_stuff_documents_chain(llm, prompt)

# 3. Create the retrieval chain
lcel_qa_interface = create_retrieval_chain(retriever, document_chain)

# 4. First question: 'in what context is Morocco mentioned in the report?'
query1 = "in what context is Morocco mentioned in the report?"
result1 = lcel_qa_interface.invoke({"input": query1})
print("Answer to first query:", result1["answer"])

# 5. Format the chat history correctly for the next query
# The history must be a list of dictionaries with 'role' and 'content'.
chat_history = [
    {"role": "user", "content": query1},
    {"role": "assistant", "content": result1["answer"]}
]

print("-" * 50)

# 6. Second question: 'give some statistics around this.'
# This query is dependent on the previous answer.

# a. Define a new prompt for the history-aware retriever
history_aware_prompt = ChatPromptTemplate.from_messages([
    MessagesPlaceholder(variable_name="chat_history"),
    ("user", "{input}"),
    ("user", "Given the above conversation, generate a search query to look up in order to get information to answer the user's question.")
])

# b. Create the history-aware retriever
history_aware_retriever = create_history_aware_retriever(llm, retriever, history_aware_prompt)

# c. Create a new final chain that uses the history-aware retriever
conversational_retrieval_chain = create_retrieval_chain(history_aware_retriever, document_chain)

# d. Run the second query with the conversational chain
# Pass both the new query and the correctly formatted chat history
query2 = "give some statistics around this."
result2 = conversational_retrieval_chain.invoke({"input": query2, "chat_history": chat_history})

print("Answer to second query:", result2["answer"])
````

#### Automation 2 - Document Classification

Let us consider an example. Say we have a lot of files which are either invoices or receipts. We want to classify and sort these documents based on their type.

Doing this is easy using GPT API.

We create simple python functions to do this.

````python
import os
import shutil
from openai import OpenAI
import pdf2image
import pytesseract

# Set up the OpenAI client once for all functions
client = OpenAI(api_key=os.environ.get("OPENAI_API_KEY"))

# This function is fine as is, but here is a more modern version
# using pathlib for better file system handling.
def list_files_only(directory_path):
    import pathlib

    if os.path.isdir(directory_path):
        # A more concise way to get a list of PDF files
        path = pathlib.Path(directory_path)
        file_list = [f.name for f in path.glob("*.pdf") if f.is_file()]
        return file_list
    else:
        return f"{directory_path} is not a directory"

def classify(file_name):
    # The OCR part is correct, but needs to handle multi-page PDFs
    detected_text = ''
    images = pdf2image.convert_from_path(file_name)
    for page in images:
        detected_text += pytesseract.image_to_string(page) + '\n\n'

    system_msg = 'You are an accounts payable expert.'

    query = '''
    Classify this document and return one of these two document types as response - [Invoices, Receipts]
    Return only the document type in the response.

    Document Type =
    '''
    user_msg = detected_text + '\n\n' + query

    # The key update: use the new client.chat.completions.create() method
    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[{"role": "system", "content": system_msg},
                  {"role": "user", "content": user_msg}]
    )

    # Use dot notation to get the content from the new response object
    return response.choices[0].message.content

# This function is fine as is
def move_file(current_path, new_folder):
    if os.path.isfile(current_path) and os.path.isdir(new_folder):
        file_name = os.path.basename(current_path)
        new_path = os.path.join(new_folder, file_name)
        shutil.move(current_path, new_path)
        print(f'File moved to {new_path}')
````

We create two folders labelled 'Invoices' & 'Receipts', in the folder where the unclassified invoices & receipts are present.

Let us execute the code now to classify these files and sort them into separate folders based on the document type.

````python
import zipfile
import os

zip_path = '/content/invoices and receipts.zip'
extract_path = '/content/documents'

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Extracted files:", os.listdir(extract_path))
````

````python
pdf_count = 0
for root, _, files in os.walk('/content/documents'):
    for f in files:
        if 'pdf' in f.lower():
            print("FOUND:", os.path.join(root, f))
            pdf_count += 1

print("Total PDFs:", pdf_count)
````

````python
def list_pdfs_recursive(folder):
    pdfs = []
    for root, _, files in os.walk(folder):
        for file in files:
            if file.lower().endswith('.pdf'):
                pdfs.append(os.path.join(root, file))
    return pdfs
````

````python
if __name__ == '__main__':
    source_folder = '/content/documents'
    invoice_folder = '/content/invoices'
    receipt_folder = '/content/receipts'

    os.makedirs(invoice_folder, exist_ok=True)
    os.makedirs(receipt_folder, exist_ok=True)

    pdf_files = list_pdfs_recursive(source_folder)

    print(f"Total PDFs found: {len(pdf_files)}")

    for current_path in pdf_files:
        classification = classify(current_path)

        print(
            f"File: {os.path.basename(current_path)} "
            f"-> Classified as: {classification}"
        )

        if "Invoices" in classification:
            move_file(current_path, invoice_folder)
        elif "Receipts" in classification:
            move_file(current_path, receipt_folder)
        else:
            print(f" Could not classify {current_path}")
````

Upon execution, the code sorts these files perfectly!

#### Automation 3 - Recipe Recommendations

We can even feed our favorite cookbooks to GPT API, and ask it to give recipe recommendations based on our inputs. Let us look at an example. We use the Brakes' Meals n More recipe cookbook, and talk to it using LangChain. Let us ask it to give recommendations based on the ingredients we have at home.

````python
# updated_retrievalqa.py
'''
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.vectorstores.faiss import FAISS
from langchain.chat_models import ChatOpenAI
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_text_splitters.character import RecursiveCharacterTextSplitter
from langchain.chains import RetrievalQA
import os
'''

# ----- config -----
PDF_PATH = "/content/meals-more-recipes.pdf"
INDEX_DIR = "index_store"
EMBEDDING_MODEL = "text-embedding-3-small"     # or other embedding model you prefer
LLM_MODEL = "gpt-4.1-mini"                     # per your preference; change if needed

# ----- load + chunk -----
loader = PyPDFLoader(PDF_PATH)
docs = loader.load()   # list[Document]

splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunks = splitter.split_documents(docs)   # list[Document]

# ----- embeddings + index build -----
emb = OpenAIEmbeddings(model=EMBEDDING_MODEL)
faiss_index = FAISS.from_documents(chunks, emb)

# persist local index
faiss_index.save_local(INDEX_DIR)

# ----- load index (example of later usage) -----
loaded = FAISS.load_local(INDEX_DIR, emb,allow_dangerous_deserialization=True)
retriever = loaded.as_retriever(search_kwargs={"k": 6})

# ----- LLM + RetrievalQA chain -----
llm = ChatOpenAI(model_name=LLM_MODEL, temperature=0.0)
qa = RetrievalQA.from_chain_type(llm=llm, chain_type="stuff", retriever=retriever, return_source_documents=True)

# ----- run query -----
query = (
    "I have a lot of broccoli and tomatoes at home. "
    "Recommend a recipe for some meal I can make at home using these."
)

# RetrievalQA.run usually returns a string answer. If return_source_documents=True you may want the dict.
# Try both ways and print available keys.
try:
    # preferred short way
    answer = qa.run(query)
    print("Answer:\n", answer)
except Exception:
    # fallback: some langchain versions return a dict when called with a dict
    result = qa({"query": query})
    print("Answer:\n", result.get("result") or result)
    if "source_documents" in result:
        print("\nSource documents (titles / page info):")
        for doc in result["source_documents"]:
            # Document has .page_content and .metadata; metadata often contains 'source' or 'page'
            print("-", doc.metadata.get("source", "unknown"), " / ", doc.metadata.get("page", "n/a"))
````

Upon execution, the PDF recommends a recipe for a meal that can be prepared using the mentioned ingredients!

We use the textbook Advanced High-School Mathematics by David B. Surowski and ask the LLM to create a question paper with a marking scheme for a particular chapter in the textbook.

We execute the below code -

````python
# retrievalqa_further.py
from dotenv import load_dotenv
load_dotenv()  # loads OPENAI_API_KEY from .env
'''
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.vectorstores.faiss import FAISS
from langchain.chat_models import ChatOpenAI
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_text_splitters.character import RecursiveCharacterTextSplitter
#from langchain.chains import RetrievalQA
import os
'''
# ----- config -----
PDF_PATH = "/content/further.pdf"
INDEX_DIR = "index_store"
EMBEDDING_MODEL = "text-embedding-3-small"
LLM_MODEL = "gpt-4.1-mini"   # change if you prefer another model
CHUNK_SIZE = 1000
CHUNK_OVERLAP = 200
K = 6

# ----- load + split -----
loader = PyPDFLoader(PDF_PATH)
docs = loader.load()  # -> list[Document]

splitter = RecursiveCharacterTextSplitter(chunk_size=CHUNK_SIZE, chunk_overlap=CHUNK_OVERLAP)
chunks = splitter.split_documents(docs)  # list[Document]

# ----- embeddings + index build (or load if index already exists) -----
emb = OpenAIEmbeddings(model=EMBEDDING_MODEL)  # requires OPENAI_API_KEY env var

if not os.path.exists(INDEX_DIR) or not os.listdir(INDEX_DIR):
    print("Building FAISS index from documents...")
    faiss_index = FAISS.from_documents(chunks, emb)
    faiss_index.save_local(INDEX_DIR)
    print("Index saved to", INDEX_DIR)
else:
    print("Index folder exists — loading index from", INDEX_DIR)

# Load the index (must pass same embeddings instance)
vector_index = FAISS.load_local(INDEX_DIR, emb,allow_dangerous_deserialization=True)
retriever = vector_index.as_retriever(search_kwargs={"k": K})

# ----- LLM + RetrievalQA -----
llm = ChatOpenAI(model_name=LLM_MODEL, temperature=0.0)
qa = RetrievalQA.from_chain_type(llm=llm, chain_type="stuff", retriever=retriever, return_source_documents=True)

# ----- query -----
query = (
    'list 5 questions of 20 marks total of varying difficulty and weightage '
    'based on the topic "Euclidian Geometry"'
)

# RetrievalQA.invoke
try:
    answer = qa.invoke(query)
    print("\nAnswer:\n", answer)
except Exception:
    result = qa({"query": query})
    # result may be {'result': '...', 'source_documents': [...]}
    print("\nAnswer:\n", result.get("result") or result)
    if "source_documents" in result:
        print("\nSource documents used:")
        for d in result["source_documents"]:
            print("-", d.metadata.get("source", "unknown"), d.metadata.get("page", "n/a"))
````

The LLM reads the PDF textbook and create the question paper for us!
